# Data Visualization with Seaborn Python Library

Seaborn is a Python library used for creating statistical data visualizations. It is built on top of Matplotlib and works closely with Pandas DataFrames, making it easier to create attractive and informative graphs with minimal code.

Features of Seaborn
* It is a library for making statistical graphics in Python.
* It is built on top of Matplotlib and integrates closely with Pandas data structures.
* It helps in exploring and understanding data through visualization.
* It can directly work with DataFrames and arrays.
* It automatically performs statistical aggregation and semantic mapping while plotting.
* It provides a high-level, declarative API, allowing users to focus on data interpretation rather than plot formatting.
* It produces beautiful and informative visualizations with simple commands.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('titanic')
df.head()
```

### Count Plot in Seaborn
A Count Plot is a type of categorical plot in Seaborn that displays the count (frequency) of observations in each category using bars.
It is similar to a histogram, but it is used for categorical data instead of numerical data.

```python
sns.countplot(x = 'sex' , data = df)
```

```python
sns.countplot(x='sex', hue = 'survived', data = df, palette = 'Set1')
```

* The hue parameter splits each bar into different colors based on another categorical variable

### KDE Plot (Kernel Density Estimation Plot) in Seaborn

A KDE (Kernel Density Estimation) Plot is a distribution plot used to visualize the probability density of a continuous variable. It provides a smooth curve that represents the distribution of data, making it an alternative to a histogram.

```python
sns.kdeplot(x = 'age' ,data=df ,color = 'black')
```

### Distribution Plot in Seaborn

A Distribution Plot is used to understand how the values of a variable are distributed. It helps identify the shape, spread, center, skewness, and presence of outliers in the data.

Seaborn provides several functions for plotting distributions

```python
sns.displot(x = 'age' , kde = True , bins = 5 , data = df)
```

```python
sns.displot(x = 'age' , kde = True , bins = 5 ,hue = df['survived'], palette='Set3',data =df)
```

### Scatter Plot in Seaborn

A Scatter Plot is a relational plot used to display the relationship between two numerical variables. Each point on the graph represents one observation in the dataset.

It helps identify:

* Correlation between variables
* Trends or patterns
* Clusters
* Outliers

```python
df = sns.load_dataset('iris')
df.head()
```

```python
sns.scatterplot(x='sepal_length' , y = 'petal_length' , data = df , hue = 'species')
```

### Joint Plot in Seaborn

A Joint Plot is a Seaborn relational/distribution plot that displays the relationship between two numerical variables along with the distribution of each variable.

It combines:

* A scatter plot (or another plot type) in the center.
* A histogram (or KDE plot) for the x-axis.
* A histogram (or KDE plot) for the y-axis.

```python
sns.jointplot(x='sepal_length' , y ='petal_length', data = df, kind = 'reg')
```

### Pair Plot in Seaborn

A Pair Plot is a Seaborn plot that displays the pairwise relationships between multiple numerical variables in a dataset. It creates a matrix of scatter plots, with histograms (or KDE plots) on the diagonal.

It is very useful for Exploratory Data Analysis (EDA) because it helps visualize relationships among all numerical features at once.

```python
sns.pairplot(df)
```

### Heatmap in Seaborn

A Heatmap is a matrix plot used to represent data values using different colors. It is commonly used to visualize correlation matrices, missing values, and numerical data in matrix form.

Darker or brighter colors indicate higher or lower values, making patterns easy to identify.

```python
corr = df.corr(numeric_only=True)

sns.heatmap(corr)

plt.show()
```

```python
sns. heatmap(corr, cmap=['red', 'green', 'blue'], vmin = -.5,
vmax = 0.6, annot = True)
```

### relplot() in Seaborn
relplot() (Relational Plot) is a figure-level function in Seaborn used to visualize the relationship between two numerical variables. It can create scatter plots and line plots and supports grouping data by color, style, size, and facets.

```python
import seaborn as sns

# Apply the default theme
sns.set_theme()

# Load the example dataset
tips = sns.load_dataset("tips")

# Create a relational plot
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    col="time",
    hue="smoker",
    style="smoker",
    size="size"
)
```

### Strip Plot in Seaborn
A Strip Plot is a categorical plot used to display the distribution of a numerical variable across one or more categorical variables. It represents each observation as an individual point.

To avoid overlapping points, Seaborn adds a small amount of random displacement called jitter.

```python
# x-axis values
x = ['sun', 'mon', 'fri', 'sat', 'tue', 'wed', 'thu']

# y-axis values
y = [5, 6.7, 4, 6, 2, 4.9, 1.8]

# Plotting strip plot
ax = sns.stripplot(x=x, y=y)

# Labels
ax.set(xlabel='Days', ylabel='Amount Spend')

# Title
plt.title('My First Strip Plot')

# Display plot
plt.show()
```

```python
# use to set style of background of plot
sns.set(style="whitegrid")

# loading data-set
iris = sns.load_dataset('iris')

# plotting strip plot with seaborn
# deciding the attributes of dataset on
# which plot should be made
ax = sns.stripplot(x='species', y='sepal_length', data=iris)

# giving title to the plot
plt.title('Graph')

# function to show plot
# plt.show()
```

### Swarm Plot in Seaborn
A Swarm Plot is a categorical plot that displays individual data points without allowing them to overlap. It is similar to a Strip Plot, but instead of using random jitter, Seaborn automatically arranges the points so that every point is visible.

```python
tips = sns.load_dataset("tips")

sns.swarmplot(data=tips, x="day", y="total_bill")

plt.show()
```

### Box Plot in Seaborn
A Box Plot is a categorical plot used to visualize the distribution of numerical data. It summarizes the data using the five-number summary and helps identify the spread, central tendency, and outliers.

```python
df = sns.load_dataset('tips')

sns.boxplot(x='day', y='total_bill', data=df, hue='smoker')
```

### Violin Plot in Seaborn

A Violin Plot is a categorical plot that combines the features of a Box Plot and a KDE (Kernel Density Estimation) Plot. It shows both the distribution and the probability density of the data.

```python
# reading the dataset
df = sns.load_dataset('tips')
sns. violinplot(x='day', y='total_bill', data=df,
hue='sex', split=True)
```
