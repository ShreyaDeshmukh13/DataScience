# Pandas Python Library

Pandas is an open-source Python library built on top of NumPy that provides fast, flexible, and expressive data structures designed for data manipulation, cleaning, and analysis.


```python
import pandas as pd
import numpy as np
```

The Pandas library relies on two primary data structures: the Series and the DataFrame .
*  Series is a one-dimensional, labeled array capable of holding any data type
* A DataFrame is a two-dimensional, tabular data structure with labeled axes (rows and columns).

## Series
A Series consists of two parts:
* Data – Stores the actual values (can be any NumPy data type).
* Index – Stores labels for each data value.
Each value in the data array is associated with a corresponding label in the index array.
The index makes it easy to identify and access individual elements
* The Series constructor is used to create a Pandas Series.
* By default, pandas will assign numerical values increasing from 0 as labels.


```python
s = pd.Series([12,-4,7,9])
s
```

```python
s = pd.Series([12,-4,7,9] , index = ['a','b','c','d'])
s
```

The two arrays that make up a Pandas Series are:

1. Data Array
* Stores the actual values of the Series.
* Can contain any NumPy-supported data type (integers, floats, strings, etc.).
2. Index Array
* Stores the labels (indexes) for each data value.
* Each label uniquely identifies the corresponding element in the data array.

```python
mydict2 = {'red' : 400 , 'yellow' : 100 , 'black' : 700}
myseries2 = pd.Series(mydict2)
myseries + myseries2
```

* When adding two Series, pandas aligns them by their index labels, not by their positions.
* red: 2000 + 400 = 2400
* yellow: 500 + 100 = 600
* blue: No matching value in myseries2 → NaN
* orange: No matching value in myseries2 → NaN
* black: No matching value in myseries → NaN

```python
import pandas as pd

mydict = {
    'red': 'Apple',
    'blue': 'Sky',
    'yellow': 'Sun',
    'orange': 'Mango'
}

myseries = pd.Series(mydict)

mydict2 = {
    'red': ' Juice',
    'yellow': ' Flower',
    'black': 'Night'
}

myseries2 = pd.Series(mydict2)

print(myseries)
print(myseries2)

print(myseries + myseries2)
```

### DataFrame

* A DataFrame is a two-dimensional, size-mutable, and heterogeneous tabular data structure with labeled rows (index) and columns. It is similar to an Excel spreadsheet or a SQL table.


```python

data = {'color' : ['blue','green','yellow','red','white'],
'object' : ['ball', 'pen','pencil', 'paper','mug'],
'price' : [1.2,1.0,0.6,0.9,1.7]}

frame = pd.DataFrame(data)
frame
```

```python
frame2 = pd.DataFrame(data,columns=['objects','price'])
frame2
```

```python
frame2 = pd.DataFrame(data,index = ['one','two','three','four','five'])
frame2 
```

The DataFrame constructor accepts three main arguments:

1. data
* Contains the actual data.
* It can be a dictionary, list, tuple, NumPy array, or another DataFrame.
2. index
* Contains the labels for the rows.
* If not specified, pandas assigns default row numbers (0, 1, 2, ...).
3. columns
* Contains the names (labels) of the columns.
* If not specified, pandas assigns default column names.

```python
frame3 = pd.DataFrame(np.arange(16).reshape((4,4)),
index = ['red','blue','yellow','white'],
columns = ['ball','pen','pencil','paper'])

frame3
```

### Operations Between DataFrame and Series


* when you perform an operation between a DataFrame and a Series, Pandas aligns the Series index with the DataFrame columns (by default).

```python
frame = pd.DataFrame(np.arange(16).reshape((4,4)),
index=['red','blue','yellow','white'],
columns=['ball','pen','pencil', 'paper'])

frame
```

```python
ser = pd.Series(np.arange(4), index=['ball','pen','pencil', 'paper'])
ser
```

```python
frame - ser
```

* Pandas matches the Series labels with the column names. It subtracts the Series from every row.


```python
ser - frame
```

* Each value in the Series is subtracted by the corresponding DataFrame value.

```python
ser['mug'] = 9
ser
```

```python
frame - ser
```

* There is no DataFrame value for mug, so it is treated as missing
