* column_stack() alias for hstack
* row_stack() alias for vstack 

```python
a = np.array([0,1,2])
b = np.array([3,4,5])
c = np.array([6,7,8])
```

```python
np.column_stack((a,b,c))
```

```python
np.row_stack((a,b,c))
```

* The AttributeError: module 'numpy' has no attribute 'row_stack' occurs because np.row_stack has been officially deprecated and completely removed from modern versions of NumPy.

* NumPy clean-up updates removed redundant and duplicate "alias" functions. Because row_stack and vstack did the exact same thing behind the scenes, the creators kept only np.vstack to make the library cleaner and faster.

* While np.row_stack was purely a "useless duplicate" of np.vstack, np.column_stack actually does something completely different from its horizontal sibling, np.hstack, when given 1D arrays

### Basic Operations on Arrays - split()
* The numpy.split() function divides a NumPy array into multiple sub-arrays along a specified axis. It requires that the array be split into equally sized parts; otherwise, it throws a ValueError

```python
array24 = np.array([[10,-7,0,20],[-5,1,200,40],[30,1,-1,4],[1,2,0,4],[0,1,0,2]])
array24
```

```python
#[1,3] indicates the row indices on which to split the array , By default , axis = 0
first , second , third = np.split(array24,[1,3])
print("first = \n" , first)
print("second = \n" , second)
print("third = \n" , third)
```

```python
first , second , third = np.split(array24,[3,4],axis = 1)
print("first = \n" , first)
print("second = \n" , second)
print("third = \n" , third)
```

```python
# 2nd parameter 2 implies array is to be split in 2 equal parts over
# axis=1 along the column axis

firsthalf , secondhalf = np.split(array24,2,axis = 1)
print("first = \n" , firsthalf)
print("second = \n" , secondhalf)
```

* np.vsplit (Vertical Split) cuts your data horizontally like a knife, slicing it into row blocks.

* np.hsplit (Horizontal Split) cuts your data vertically like a knife, slicing it into column blocks.

* np.dsplit (Depth Split) slices 3D arrays along the third axis (depth)

```python
a = np.arange(16.0).reshape(4,4)
a
```

```python
np.hsplit(a,2)
```

```python
np.hsplit(a,np.array([2,3]))
```

```python
b = np.arange(20.0).reshape(4,5)
b
```

```python
np.vsplit(b,2)
```

```python
np.vsplit(b, np.array([1, 3]))
```

```python
c = np.arange(12.0).reshape (2,2,3)
c
```

```python
np.dsplit(c,3)
```

```python
np.dsplit(c, np.array([1,3]))
```
